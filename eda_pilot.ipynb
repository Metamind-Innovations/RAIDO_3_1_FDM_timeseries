```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def plot_clients(client_paths: list[str]):
    colors = ["blue", "orange"]
    fig, axes = plt.subplots(1, len(client_paths), figsize=(10 * len(client_paths), 4), sharey=True)
    if len(client_paths) == 1:
        axes = [axes]
    for ax, path, color in zip(axes, client_paths, colors):
        df = pd.read_csv(path)
        df["f_time"] = pd.to_datetime(df["f_time"], utc=True)
        client_name = path.split("/")[-1].replace(".csv", "")
        ax.plot(df["f_time"], df["n_value"], color=color, linewidth=0.8)
        ax.set_title(client_name)
        ax.set_xlabel("Time")
        ax.set_ylabel("Value")
        ax.tick_params(axis="x", rotation=30)
    plt.suptitle("Client Data", fontsize=14)
    plt.tight_layout()
    plt.show()

def plot_synthetic(synth_path: str):
    df = pd.read_csv(synth_path)
    df["f_time"] = pd.to_datetime(df["f_time"])
    fig, ax = plt.subplots(figsize=(12, 4))
    ax.plot(df["f_time"], df["n_value"], color="green", linewidth=0.8)
    ax.set_title("Synthetic Data")
    ax.set_xlabel("Time")
    ax.set_ylabel("Value")
    ax.tick_params(axis="x", rotation=30)
    plt.tight_layout()
    plt.show()
```
